## Building Effective Agents with LangChain4j

As [mentioned here](https://mirror-feeling-d80.notion.site/Workflow-And-Agents-17e808527b1780d792a0d934ce62bee6), building agents does not require a specific framework.
However, using a framework can make the process much easier.
The concept of an agent can be implemented with any framework.
In this example, I am using LangChain4j.

### Setup Dependency

Please download this dependency first:

```kotlin
@file:DependsOn("dev.langchain4j:langchain4j:1.0.1")
@file:DependsOn("dev.langchain4j:langchain4j-ollama:1.0.1-beta6")
```

### Setting Up the AI Model

In LangChain4j, the `ChatLanguageModel` serves as an abstraction for AI models.
You can use it with various providers, such as the OpenAI API, Anthropic API, AWS Bedrock, or Ollama.
In this example, we will use Ollama.

```kotlin
import dev.langchain4j.model.ollama.OllamaChatModel
import dev.langchain4j.model.chat.ChatModel

val modelName = "llama3.2"
val builder = OllamaChatModel.builder()
    .baseUrl("http://localhost:11434")
    .temperature(0.0)
    .modelName(modelName)
val llm: ChatModel = builder.build()

val answer = llm.chat("Please says 'Hello, World!' in Spanish")
println(answer) // Hola Mundo!
```

## Augmented LLM

Great! Now that we have completed the necessary prerequisites, we can start using agent patterns.
We will begin with an Augmented LLM that incorporates search and tools.

### Search Example

```kotlin
import dev.langchain4j.service.AiServices
import com.fasterxml.jackson.annotation.JsonCreator
import com.fasterxml.jackson.annotation.JsonProperty

data class SearchQuery(
    val searchQuery: String = "",
    val justification: String = ""
)

interface SearchQueryService {
    fun generateSearchQuery(userQuestion: String): SearchQuery
}

val service = AiServices.builder(SearchQueryService::class.java)
    .chatModel(llm)
    .build()

val result = service.generateSearchQuery("How does Calcium CT score relate to high cholesterol?")

println(result.searchQuery) // Calcium CT score and high cholesterol
println(result.justification) // A Calcium CT score is a measure of calcium deposits in the arteries, which..
```
